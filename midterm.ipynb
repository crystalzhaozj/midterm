```python
# Define the function "calculate_days"
import datetime
def calculate_days(str):
    temp_list = str.split(':')[1].split('-')
    start = temp_list[0].strip()
    end = temp_list[1].strip()
    start_date = datetime.datetime.strptime(start, '%b %d, %Y')
    end_date = datetime.datetime.strptime(end, '%b %d, %Y')
    interval = end_date-start_date
    return interval.days 
```

```python
phrase = 'Aired: Jun 18, 2022 - Aug 28, 2022'
# test that your function works

calculate_days(phrase)
```
